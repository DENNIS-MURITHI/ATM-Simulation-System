# ATM Simulation System

## Overview
A secure, user-friendly Automated Teller Machine (ATM) simulation designed to demonstrate fundamental banking operations and account management principles. This application enables users to perform essential financial transactions after successful authentication.

## Features

### 🔐 Security
- **PIN Authentication**: Users must enter a valid PIN to access the system
- **Account Protection**: Unauthorized access attempts are blocked with appropriate error messaging

### 💰 Core Operations
- **Balance Inquiry**: Check current account balance at any time
- **Withdrawal**: Withdraw funds with real-time balance validation
- **Deposit**: Add funds to account balance
- **Exit**: Safely terminate the session

## Technical Specifications

### System Requirements
- Python 3.x
- Console/Terminal environment

### Initial Configuration
- **Default PIN**: 1234
- **Initial Balance**: 40,000

### Transaction Validation
- Withdrawal requests are validated against available balance
- Insufficient fund alerts prevent negative transactions
- All transactions display updated balance confirmation

## Usage Flow

```
1. Launch Application
   ↓
2. Enter PIN
   ├─→ Incorrect PIN → Error Message
   └─→ Correct PIN → Continue
   ↓
3. Select Operation (1-4)
   ├─→ Option 1: Display Balance
   ├─→ Option 2: Withdraw Funds
   ├─→ Option 3: Deposit Funds
   └─→ Option 4: Exit Program
   ↓
4. Perform Transaction
   ↓
5. Display Updated Balance
```

## Error Handling
- **Invalid PIN**: System denies access with confirmation message
- **Insufficient Balance**: Withdrawal rejected with balance check
- **Invalid Menu Selection**: User can retry selection
